## The Python Data Model & Class Fundamentals
### 1. Objects, Values, and Types
"Objects are Python’s abstraction for data. All data in a Python program is represented by objects or by relations between objects."

This means integers, strings, functions, and even classes themselves are objects. Every object in Python has exactly three things:

Identity: The object's address in memory. Once an object is created, its identity never changes. You check this using the id() function, or compare identity using the is operator.

Type: What kind of object it is (e.g., int, str, list, or your custom class). The type determines what operations the object supports. You check this using the type() function.

Value: The actual data the object holds.

Mutability vs. Immutability
An object's type determines if its value can change.

Immutable: The value cannot change once created (e.g., int, float, str, tuple). If you "change" a string, Python actually creates a brand-new string object with a new identity.

Mutable: The value can be changed in place (e.g., list, dict, set, and most custom objects we will build). Its identity remains the same while the contents change.

### 2. Class and Instance Definition
If built-in objects like str and list are standard data abstractions, a Class allows you to build your own custom data abstractions.

Class: The blueprint. It defines the structure and behaviors.

Instance (Object): A concrete realization of that blueprint.

Let's look at the basic syntax. We will build a simple system for a university.

````python
# The 'class' keyword defines the blueprint.
# By convention, class names use PascalCase (CamelCase starting with a capital letter).
class Student:
    """A blueprint representing a university student."""
    pass

# We 'instantiate' the class by calling it like a function.
# This creates a unique object in memory.
student_1 = Student()
student_2 = Student()

print(type(student_1))       # Output: <class '__main__.Student'>
print(id(student_1))         # Output: 14032483... (Memory address)
print(student_1 is student_2) # Output: False (They are distinct objects)
````

### 3. The Anatomy of self and __init__
Right now, our Student objects are empty. We need to give them state (values).

When you instantiate a class (e.g., Student()), Python automatically calls a special method named __init__ to initialize the newly created object.

````python
class Student:
    # __init__ is the initializer (often called the constructor).
    # 'self' is required as the first parameter in almost all instance methods.
    def __init__(self, first_name, last_name, roll_number):
        # We attach data to the specific instance using 'self.attribute_name = value'
        self.first_name = first_name
        self.last_name = last_name
        self.roll_number = roll_number
        self.enrolled = True

# When we call Student(), Python creates a blank object in memory.
# It then silently passes that new object as the 'self' parameter to __init__,
# along with the other arguments we provide.
alice = Student("Alice", "Smith", "CS-101")
````

#### The golden rule of self:
self represents the specific instance of the class that is currently being operated on.
When you write alice.first_name, Python looks at the object alice and fetches the first_name attribute you attached to it via self.

````python
class Student:
    def __init__(self, first_name, last_name, roll_number):
        self.first_name = first_name
        self.last_name = last_name
        self.roll_number = roll_number
    
    # Instance method: 'self' is always the first parameter
    def get_full_name(self):
        # We must use 'self.' to access the instance's attributes
        return f"{self.first_name} {self.last_name}"

bob = Student("Bob", "Jones", "CS-102")

# We call the method on the instance.
print(bob.get_full_name()) # Output: Bob Jones
````

```text
Here is the crucial part that explains why self exists: When you type bob.get_full_name(), Python translates that under the hood into:
Student.get_full_name(bob)
Python takes the instance (bob) and implicitly passes it in as the first argument (self). That is why self must be defined in the method signature, even though you don't pass it when calling the method.

### 4. Namespaces and __dict__

Where does Python actually store self.first_name?
Behind the scenes, almost every custom object is backed by a dictionary. This dictionary is the object's namespace. You can view it using the special __dict__ attribute.

````python
charlie = Student("Charlie", "Brown", "CS-103")
    
# Let's peek under the hood
print(charlie.__dict__) 
# Output: {'first_name': 'Charlie', 'last_name': 'Brown', 'roll_number': 'CS-103'}

# Because it's just a dictionary underneath, Python allows us to add attributes dynamically!
charlie.major = "Computer Science"

print(charlie.__dict__)
# Output: {'first_name': 'Charlie', 'last_name': 'Brown', 'roll_number': 'CS-103', 'major': 'Computer Science'}
````

This dynamic nature is powerful, but it's generally best practice to define all expected attributes within __init__ so anyone reading your code knows exactly what state the object should have.
