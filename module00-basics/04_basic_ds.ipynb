# Phase 3: Core Data Structures

Data structures are specialized formats for organizing, storing, and retrieving data in memory. Python provides four primary built-in data structures. Choosing the right one depends on two main concepts: **mutability** (can it be changed after creation?) and **ordering** (does it maintain the sequence of its elements?).

### 1. Lists `[]`
* **Properties:** Ordered, **Mutable** (changeable), allows duplicate elements.
* **Under the hood:** Python lists are dynamic arrays. They allocate extra contiguous memory in advance, which is why appending items is fast, but it makes them slightly heavier in memory.
* **Best for:** Storing sequences of items where you need to frequently add, remove, sort, or modify elements.
* **Indexing & Slicing:** Elements are accessed via zero-based indices. Slicing `[start:stop:step]` allows you to extract sub-lists easily.

### 2. Tuples `()`
* **Properties:** Ordered, **Immutable** (cannot be changed once created), allows duplicates.
* **Under the hood:** Because their size is fixed upon creation, tuples are highly memory-efficient and slightly faster than lists. 
* **Best for:** "Write-protecting" data that should never change throughout the program's lifecycle (e.g., RGB color codes, geographic coordinates). They are also the standard way to return multiple values from a function.

### 3. Dictionaries `{key: value}`
* **Properties:** Key-Value pairs, Ordered (insertion order is preserved as of Python 3.7+), Mutable. 
* **Rules:** Keys must be strictly **unique** and **immutable** (strings, integers, or tuples). Values can be anything (including other lists or dictionaries).
* **Under the hood:** Implemented as Hash Maps. Looking up a value by its key takes the exact same amount of time whether the dictionary has ten items or ten million items (O(1) time complexity).
* **Best for:** Storing structured data (like JSON or API responses), modeling real-world entities (e.g., a user profile), or lightning-fast lookups.

### 4. Sets `{}`
* **Properties:** Unordered, Mutable, **Strictly Unique** elements (duplicates are automatically dropped).
* **Under the hood:** Essentially a dictionary with only keys and no values. This structure makes "membership testing" (checking if `x in set`) blazing fast compared to lists.
* **Best for:** Instantly removing duplicates from a dataset, or performing mathematical operations like Intersections (finding common items) and Unions (combining distinct items).

# Deep Dive: List and Dictionary Comprehensions

Comprehensions are one of Python's defining features. They provide a concise, readable, and computationally efficient way to create new lists, dictionaries, or sets based on existing iterables. 

Writing "Pythonic" code often means replacing multi-line `for` loops containing `.append()` statements with a single, elegant comprehension. Because comprehensions are optimized in Python's underlying C code, they execute faster than standard loops.

### 1. Anatomy of a List Comprehension
The basic formula for a list comprehension is enclosed in square brackets `[]`:
`[expression for item in iterable if condition]`

* **`expression`**: What you want to append to the new list (can be modified or computed).
* **`for item in iterable`**: The standard loop.
* **`if condition` (Optional)**: A filter. If this evaluates to False, the item is skipped.

### 2. The `If/Else` Trap
A common point of confusion is where to put the `if` statement:
* **Filtering:** If you want to *skip* items, the `if` goes at the **very end**.
* **Transforming:** If you want to keep all items but *change* their value based on a condition (an if/else), the logic goes at the **very beginning** (in the expression slot).

### 3. Anatomy of a Dictionary Comprehension
Dictionary comprehensions follow the exact same logic, but they use curly braces `{}` and require a key-value pair separated by a colon in the expression slot:
`{key_expression: value_expression for item in iterable if condition}`

### 4. When NOT to use them
While comprehensions are powerful, do not sacrifice readability for cleverness. If you have nested loops with multiple complex conditions, a standard `for` loop is better. If a comprehension spans more than two lines, it's usually too complex.

```python
# ==========================================
# 1. BASIC LIST COMPREHENSION VS. FOR LOOP
# ==========================================
prices = [10, 20, 30, 40]

# The Old Way (Standard Loop):
taxed_prices_loop = []
for price in prices:
    taxed_prices_loop.append(price * 1.2)

# The Pythonic Way (List Comprehension):
# Read it as: "Give me price * 1.2 for every price in prices"
taxed_prices_comp = [price * 1.2 for price in prices]
print(f"Taxed Prices: {taxed_prices_comp}")

# ==========================================
# 2. FILTERING WITH 'IF' (At the end)
# ==========================================
# Extract only the even numbers from a range
evens_only = [num for num in range(1, 11) if num % 2 == 0]
print(f"Evens: {evens_only}")

# ==========================================
# 3. TRANSFORMING WITH 'IF/ELSE' (At the beginning)
# ==========================================
# Categorize numbers as "Even" or "Odd"
# Read it as: "Give me 'Even' if num % 2 == 0, else give me 'Odd'..."
categories = ["Even" if num % 2 == 0 else "Odd" for num in range(1, 6)]
print(f"Categories: {categories}")

# ==========================================
# 4. DICTIONARY COMPREHENSIONS
# ==========================================
names = ["Alice", "Bob", "Charlie"]

# Create a dictionary where the key is the name, and the value is the length of the name
name_lengths = {name: len(name) for name in names}
print(f"\nName Lengths: {name_lengths}")

# Practical Example: Flipping a dictionary's keys and values
# Note: This only works safely if the original values are strictly unique!
user_ids = {"alice123": 1, "bob456": 2, "charlie789": 3}
id_to_username = {value: key for key, value in user_ids.items()}
print(f"Flipped Dict: {id_to_username}")

# ==========================================
# 5. NESTED LOOPS IN COMPREHENSIONS (Advanced)
# ==========================================
# Flattening a 2D matrix (a list of lists) into a single 1D list
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# The order of the loops reads identically to how you would write nested for-loops:
# "for row in matrix -> for num in row"
flat_list = [num for row in matrix for num in row]
print(f"\nFlattened Matrix: {flat_list}")

# ==========================================
# 6. SET COMPREHENSIONS (Bonus)
# ==========================================
# Just like list comprehensions, but with curly braces. Automatically removes duplicates.
words = ["apple", "banana", "apple", "cherry", "banana"]
unique_first_letters = {word[0].upper() for word in words}
print(f"Unique First Letters: {unique_first_letters}")
```
