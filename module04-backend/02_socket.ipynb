### Before we use Uvicorn and FastAPI, let's build a raw socket server to see exactly what they abstract away from us.

```python
import socket

def run_raw_server():
    # 1. Create a raw OS socket (IPv4, TCP stream)
    server_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    
    # 2. Bind the socket to localhost on port 8080
    server_socket.bind(('127.0.0.1', 8080))
    server_socket.listen(1)
    print("Listening for TCP connections on port 8080...")
    
    # 3. Accept the incoming TCP Handshake
    client_connection, client_address = server_socket.accept()
    print(f"TCP Handshake complete with: {client_address}")
    
    # 4. Read the raw HTTP bytes sent by the client
    raw_http_request = client_connection.recv(1024).decode('utf-8')
    print("\n--- RAW HTTP RECEIVED ---")
    print(raw_http_request)
    
    # 5. Send a raw HTTP response back
    http_response = (
        "HTTP/1.1 200 OK\r\n"
        "Content-Type: text/plain\r\n\r\n"
        "Hello from the raw socket!"
    )
    client_connection.sendall(http_response.encode('utf-8'))
    client_connection.close()

# Note: Running this will block the cell. You can navigate to http://127.0.0.1:8080 
# in your browser while it runs to see the raw HTTP text your browser generates!
# run_raw_server()
```
