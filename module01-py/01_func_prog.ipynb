# 1) First-Class Functions and Scoping (LEGB)

Python uses the LEGB rule, checking namespaces in this exact order:

1) Local: Variables assigned within a function.

2) Enclosing: Variables in the local scope of any enclosing functions (this is how closures work).

3) Global: Variables assigned at the top level of a module.

4) Built-in: Names preassigned in Python (like print, len).

Because functions are objects, they can "capture" variables from their enclosing scope. This creates a Closure.

```python
def make_multiplier(factor):
    # 'factor' is in the Enclosing scope of the inner function
    
    def multiplier(number):
        # The inner function remembers 'factor' even after make_multiplier finishes
        return number * factor
        
    return multiplier

# We assign the returned function object to a variable
double = make_multiplier(2)
triple = make_multiplier(3)

print(double(5))  # Output: 10
print(triple(5))  # Output: 15
```

# 2. Lambdas and Higher-Order Functions

A higher-order function is simply a function that takes another function as an argument or returns one. Python provides built-in higher-order functions that work perfectly with lambdas (anonymous, single-expression functions).

* map(function, iterable): Applies the function to every item in the iterable.

* filter(function, iterable): Keeps items where the function returns True.

* reduce(function, iterable): (From the functools module) Cumulatively applies the function to items, reducing the iterable to a single value.

```python
from functools import reduce

data = [1, 2, 3, 4, 5]

# Lambda syntax: lambda arguments: expression
# MAP: Square every number
squared = list(map(lambda x: x**2, data))
# [1, 4, 9, 16, 25]

# FILTER: Keep only even numbers
evens = list(filter(lambda x: x % 2 == 0, data))
# [2, 4]

# REDUCE: Multiply all numbers together cumulatively
# Step 1: 1*2=2, Step 2: 2*3=6, Step 3: 6*4=24, Step 4: 24*5=120
product = reduce(lambda x, y: x * y, data)
# 120
```

Key insight: While map and filter return iterator objects in Python 3 (which is why we wrapped them in list() above to see the results), they process items lazily, saving memory when dealing with massive datasets.

# 3. Comprehensions

While map and filter are powerful, Python developers generally prefer comprehensions for readability. A comprehension allows you to generate new lists, dictionaries, or sets in a single, highly optimized line of C-level execution.

The syntax pattern is: [expression for item in iterable if condition]

```python
users = [
    {"name": "Alice", "active": True, "score": 85},
    {"name": "Bob", "active": False, "score": 92},
    {"name": "Charlie", "active": True, "score": 78}
]

# List Comprehension (Replacing Map + Filter)
# Get the names of active users only
active_names = [u["name"] for u in users if u["active"]]
# ['Alice', 'Charlie']

# Dictionary Comprehension
# Create a mapping of user names to their scores
user_scores = {u["name"]: u["score"] for u in users}
# {'Alice': 85, 'Bob': 92, 'Charlie': 78}

# Set Comprehension (Notice the curly braces, but no key-value pairs)
# Get unique scores
unique_scores = {u["score"] for u in users}
# {85, 92, 78}
```

# 4. The Power of itertools

For advanced functional pipelines, the built-in itertools module is a core toolkit. It provides fast, memory-efficient tools for looping.

* chain(*iterables): Links multiple iterables together as if they were a single sequence.

* combinations(iterable, r): Returns all possible combinations of length r from the iterable.

* groupby(iterable, key_func): Groups consecutive items that share the same key. (Note: the iterable must be sorted by the key first).

```python
import itertools

# Chain
combined = list(itertools.chain([1, 2], ['a', 'b']))
# [1, 2, 'a', 'b']

# Combinations
combos = list(itertools.combinations(['A', 'B', 'C'], 2))
# [('A', 'B'), ('A', 'C'), ('B', 'C')]
```
