# Phase 2: Control Flow and Logic

Control flow dictates the order in which your code executes. Without control flow, a script runs strictly from top to bottom. With it, your program can make decisions, skip code, or repeat actions based on dynamic conditions.

### 1. Conditional Logic (`if`, `elif`, `else`)
Python evaluates conditions to `True` or `False`. 
* `if` triggers the first block of code if its condition is true.
* `elif` (Else-If) checks a new condition if the previous ones failed. You can chain as many `elif` statements as needed.
* `else` acts as the catch-all if every preceding condition was false.

### 2. "Truthiness" and "Falsiness"
A unique and powerful feature in Python is that **every object has an inherent boolean value**. You don't always need to explicitly check `if x == True:` or `if len(my_list) > 0:`.
* **Falsy values:** `0`, `0.0`, `""` (empty string), `None`, and empty collections like `[]` or `{}`.
* **Truthy values:** Almost everything else (any non-zero number, any string with text).

### 3. Iteration: `for` loops
Unlike C or Java where a `for` loop primarily counts numbers, Python's `for` loop is a **"for-each"** loop. It is designed to iterate directly over items in a sequence (like characters in a string, or items in a list).
* **The `range()` function:** Generates a sequence of numbers on the fly. `range(start, stop, step)`. Note that the `stop` number is **exclusive** (it stops *before* reaching it).

### 4. Iteration: `while` loops
A `while` loop runs continuously as long as a specified condition remains `True`. It is useful when you don't know exactly how many times you need to loop (e.g., waiting for a user to input a valid password).

### 5. Loop Control Statements
* `break`: Instantly exits the entire loop, skipping any remaining iterations.
* `continue`: Instantly skips the rest of the *current* iteration and jumps back to the top of the loop for the next one.
* `pass`: Does absolutely nothing. It acts as a syntactic placeholder when Python requires an indented block of code, but you haven't written it yet.

```python
# ==========================================
# 1. CONDITIONALS & LOGICAL OPERATORS
# ==========================================
temperature = 22
is_raining = True

if temperature < 15:
    print("Wear a heavy jacket.")
elif temperature <= 25 and is_raining:
    print("Wear a light jacket and bring an umbrella.")
elif temperature <= 25 and not is_raining:
    print("Perfect weather, just a t-shirt!")
else:
    print("It's hot out there.")

# ==========================================
# 2. EVALUATING "TRUTHINESS"
# ==========================================
user_input = ""  # An empty string is inherently "Falsy"

# Instead of checking: if user_input == "":
if not user_input:
    print("User did not type anything.")

items_in_cart = 3 # Non-zero numbers are "Truthy"
if items_in_cart:
    print(f"You have {items_in_cart} items ready for checkout.")

# ==========================================
# 3. FOR LOOPS & THE RANGE FUNCTION
# ==========================================
# range(start, stop) -> Stops before 5
print("Counting with range:")
for i in range(1, 5): 
    print(i, end=" ") # end=" " prints on the same line
print("\n") # Adds a blank line

# range(start, stop, step) -> Count backwards
print("Countdown:")
for i in range(10, 0, -2):
    print(i, end=" ")
print("\n")

# Iterating directly over a sequence (a string)
word = "Python"
print("Spelling it out:")
for letter in word:
    print(f"-> {letter}")

# ==========================================
# 4. WHILE LOOPS & CONTROL STATEMENTS
# ==========================================
counter = 0

print("\nWhile loop with controls:")
while counter < 10:
    counter += 1
    
    # Skip the number 3
    if counter == 3:
        print("Skipping 3!")
        continue 
    
    # Stop completely when we hit 6
    if counter == 6:
        print("Hit 6, breaking the loop!")
        break
        
    print(f"Processing number: {counter}")

# ==========================================
# 5. THE PASS STATEMENT
# ==========================================
def feature_in_development():
    # I haven't written this function yet, but Python requires an indented block.
    # If I leave it blank, Python throws an IndentationError.
    pass

feature_in_development() # Does nothing, but doesn't crash the program.
```
