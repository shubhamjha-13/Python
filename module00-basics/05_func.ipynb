# Functions, Scope, `*args`, and `**kwargs`

Functions are the building blocks of reusable code. They allow you to encapsulate logic, adhere to the DRY (Don't Repeat Yourself) principle, and make your programs modular. 

### 1. Anatomy of a Function
* **`def` keyword:** Used to declare a function.
* **Parameters vs. Arguments:** Parameters are the variables defined in the function signature. Arguments are the actual data passed in when the function is called.
* **Default Arguments:** You can assign a default value to a parameter, making it optional during the function call.
* **Return statement:** Functions can return data back to the caller. If no `return` is specified, Python implicitly returns `None`.

### 2. Variable Scope (The LEGB Rule)
Scope determines where in your code a variable can be accessed. Python resolves variables using the **LEGB** hierarchy (checking from inner to outer):
1. **L**ocal: Inside the current function.
2. **E**nclosing: Inside enclosing functions (if dealing with nested functions).
3. **G**lobal: Declared at the top level of the script.
4. **B**uilt-in: Python's pre-defined names (like `print`, `len`).

*Note: You can read global variables inside a function, but to modify them, you must explicitly use the `global` keyword (though this is generally considered bad practice).*

### 3. Flexible Arguments: `*args` and `**kwargs`
Sometimes you don't know exactly how many arguments a user will pass to your function. 
* **`*args` (Positional Arguments):** The single asterisk `*` gathers any number of extra positional arguments and packs them into a **Tuple**. 
* **`**kwargs` (Keyword Arguments):** The double asterisk `**` gathers any extra named arguments (e.g., `color="red", size="large"`) and packs them into a **Dictionary**.
* *Fun fact:* The words `args` and `kwargs` are just standard conventions. The magic is in the asterisks `*` and `**`. You could technically write `*numbers` or `**options`.

### 4. The Golden Rule of Argument Order
If you combine all these elements in a single function, Python enforces a strict order:
`Standard parameters -> *args -> Default parameters -> **kwargs`

```python
# ==========================================
# 1. BASIC FUNCTIONS & DEFAULT PARAMETERS
# ==========================================
# 'greeting' has a default value. It's optional when calling the function.
def greet_user(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet_user("Alice"))             # Uses default greeting
print(greet_user("Bob", "Welcome"))    # Overrides default greeting

# ==========================================
# 2. VARIABLE SCOPE (Local vs Global)
# ==========================================
system_status = "Online" # Global Scope

def update_status():
    system_status = "Offline" # Local Scope (Creates a NEW local variable)
    print(f"Inside function, status is: {system_status}")

update_status()
print(f"Outside function, status is still: {system_status}")

def force_global_update():
    global system_status # Explicitly grabs the global variable
    system_status = "Maintenance Mode"

force_global_update()
print(f"After global update, status is: {system_status}\n")

# ==========================================
# 3. *ARGS (Tuple Unpacking)
# ==========================================
# Accepts any number of numeric arguments and calculates their sum
def calculate_total(*args):
    # args is treated as a tuple: e.g., (10, 20, 30)
    total = sum(args) 
    return f"Total across {len(args)} items: ${total}"

print(calculate_total(15, 25))
print(calculate_total(100, 200, 300, 400, 500))

# ==========================================
# 4. **KWARGS (Dictionary Unpacking)
# ==========================================
# Accepts any number of named attributes to build a profile
def create_profile(username, **kwargs):
    # Start with a base dictionary
    profile = {"user": username}
    
    # kwargs is a dictionary containing all extra named arguments
    # We can merge it into our profile dictionary
    profile.update(kwargs)
    return profile

user1 = create_profile("dev_guru", role="Admin", theme="Dark", notifications=True)
print(f"\nUser Profile: {user1}")

# ==========================================
# 5. BRINGING IT ALL TOGETHER (The Order)
# ==========================================
# Order: standard (action), *args (items), default (tax_rate), **kwargs (metadata)
def process_order(action, *items, tax_rate=0.08, **metadata):
    print(f"\n--- Processing Order ---")
    print(f"Action: {action}")
    print(f"Items to process: {items}") # tuple
    print(f"Applied Tax Rate: {tax_rate}")
    
    for key, value in metadata.items():
        print(f"Metadata - {key}: {value}")

process_order(
    "PURCHASE",           # standard
    "Laptop", "Mouse",    # *args
    tax_rate=0.05,        # default (overridden)
    customer="Acme Corp", # **kwargs
    priority="High"       # **kwargs
)
```
