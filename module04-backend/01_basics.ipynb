```text
You type URL
     ↓
Browser understands the URL
     ↓
DNS finds the server's IP address
     ↓
TCP connection is established
     ↓
TLS handshake happens (for HTTPS)
     ↓
HTTP request is sent
     ↓
Server processes the request
     ↓
Database / services may be called
     ↓
HTTP response comes back
     ↓
Browser renders the result

------------------------------------------------------------------

Suppose you enter: https://example.com:443/products?id=10#reviews

```text
https://example.com:443/products?id=10#reviews
│       │           │   │         │
│       │           │   │         └── Fragment
│       │           │   └──────────── Query parameter
│       │           └──────────────── Path
│       └──────────────────────────── Domain
└──────────────────────────────────── Protocol

```text
Protocol-
http   → normal HTTP
https  → HTTP + encryption using TLS

Domain-
example.com

Humans remember names.This is where DNS comes in.

example.com
        ↓
93.184.216.34

```text
Computer
   │
   ▼
Browser Cache
   │
   │ miss
   ▼
OS Cache
   │
   │ miss
   ▼
Recursive Resolver
   │
   │ cache miss
   ▼
Root DNS
   │
   │ "Where is .com?"
   ▼
.com TLD
   │
   │ "Where is example.com?"
   ▼
Authoritative DNS
   │
   │ "93.184.216.34"
   ▼
Recursive Resolver
   │
   │ cache
   ▼
Computer
   │
   │ gets IP
   ▼
93.184.216.34

# TCP — Establishing a connection

For traditional HTTP/1.1 and HTTP/2 connections, HTTP commonly runs on top of TCP.

Think of the layers roughly like this:
```text
HTTP
 ↓
TLS        ← for HTTPS
 ↓
TCP
 ↓
IP
 ↓
Internet
```
TCP provides a reliable connection between two machines.

Before sending application data, TCP typically establishes a connection using the three-way handshake:
```text
Client                     Server

   SYN  ─────────────────►
         "Can we connect?"

        ◄─────────────────  SYN + ACK
                           "Yes"

   ACK  ─────────────────►
         "Great"
```
```text
After this:

Client  ═══════════════════════  Server
              TCP Connection
```
Now both sides can exchange data reliably.

# What is a port?
```text
An IP address identifies a machine.

A port identifies a particular network service or process on that machine.

For example:

IP address: 93.184.216.34

That tells us:

Which machine?

But this:

93.184.216.34:__443__

tells us:

Which machine + which service?

In backend development, when you run something like FastAPI:

uvicorn main:app --port 8000

you are essentially saying:

Run my application and listen for incoming connections on port 8000.

So your application might be reachable as:

http://localhost:8000

Here:

localhost → your own machine
8000      → the port where your backend is listening
